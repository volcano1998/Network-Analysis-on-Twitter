```python
import pandas as pd 
import numpy as np
```

# Load data

```python
df_covid_agg=pd.read_csv("data/user_type.csv")
node_dict = dict(zip(df_covid_agg['User ID'],df_covid_agg['y']))
all_nb = pd.read_csv("data/all_neighbor.csv")
df_mg = df_covid_agg.merge(all_nb,left_on='User ID',right_on='neighbor')
df_ct =df_mg.groupby(['screen_name','y']).count().reset_index()
intersection = set(df_mg[df_mg.group=='right'].neighbor)&set(df_mg[df_mg.group=='tagged hate'].neighbor)
```

# Overall statistics

In here, you will see the aggregated results of proportion of hate, neutral, counter-hate users in for each group.

First is right group

```python
df_mg_agg = df_mg.groupby(['group','y']).count().reset_index()
df_right = df_mg_agg[df_mg_agg['group']=='right']
df_right['total']=df_right['User ID'].sum()
df_right['perct']=df_right['User ID']/df_right['User ID'].sum()*100
df_right
```

Second is hate group

```python
df_right = df_mg_agg[df_mg_agg['group']=='tagged hate']
df_right['total']=df_right['User ID'].sum()
df_right['perct']=df_right['User ID']/df_right['User ID'].sum()*100
df_right
```

Third is the common neighbors of rigth and hate groups

```python
df_int = pd.DataFrame({"User ID": list(intersection)})
df_int = df_int.merge(df_covid_agg)
df_int = df_int.groupby('y').count().reset_index()
df_int['total']=df_int['User ID'].sum()
df_int['perct']=df_int['User ID']/df_int['User ID'].sum()*100
df_int
```

# Per account statistics

In here, you will see the proportion of hate, neutral, counter-hate neighbors every account has.

First, in right group

```python
df_mg_agg = df_mg[df_mg.group=='right'].groupby(['screen_name','y']).count().reset_index()
df_allnb = df_mg[df_mg.group=='right'].groupby(['screen_name']).count().reset_index()\
[['screen_name','y']].rename(columns={'y':'all_neighbors'})
df_mg_agg = df_mg_agg.merge(df_allnb)
df_mg_agg['neighbor type']=df_mg_agg['y'].map({-1:'hate',0:'neutral',1:'counter-hate'})
df_mg_agg['perct']=np.round(df_mg_agg['id']/df_mg_agg['all_neighbors']*100,2)
df_mg_agg.pivot(index='screen_name',columns='neighbor type',values='perct').sort_values('hate',ascending=False)
```

Second, in hate group

```python
df_mg_agg = df_mg[df_mg.group=='tagged hate'].groupby(['screen_name','y']).count().reset_index()
df_allnb = df_mg[df_mg.group=='tagged hate'].groupby(['screen_name']).count().reset_index()\
[['screen_name','y']].rename(columns={'y':'all_neighbors'})
df_mg_agg = df_mg_agg.merge(df_allnb)
df_mg_agg['neighbor type']=df_mg_agg['y'].map({-1:'hate',0:'neutral',1:'counter-hate'})
df_mg_agg['perct']=np.round(df_mg_agg['id']/df_mg_agg['all_neighbors']*100,2)
df_mg_agg.pivot(index='screen_name',columns='neighbor type',values='perct').sort_values('hate',ascending=False)
```
